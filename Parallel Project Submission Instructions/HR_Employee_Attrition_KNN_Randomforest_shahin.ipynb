## 7.1 Feature Engineering

Feature engineering creates new variables from existing features to provide additional information that may be useful for analysis and machine learning.

Two meaningful features are created:

- **TenureRatio:** Represents the proportion of an employee's total working experience spent at the current company.
- **IncomePerYearExperience:** Represents monthly income relative to the employee's total working experience.

```python
# Create TenureRatio

df["TenureRatio"] = (
    df["YearsAtCompany"] /
    df["TotalWorkingYears"].replace(0, np.nan)
)

# Replace undefined values caused by zero experience
df["TenureRatio"] = df["TenureRatio"].fillna(0)

# Display the new feature
df[[
    "YearsAtCompany",
    "TotalWorkingYears",
    "TenureRatio"
]].head()
```

```python
# Create IncomePerYearExperience

df["IncomePerYearExperience"] = (
    df["MonthlyIncome"] /
    df["TotalWorkingYears"].replace(0, np.nan)
)

# Replace undefined values caused by zero experience
df["IncomePerYearExperience"] = df[
    "IncomePerYearExperience"
].fillna(0)

# Display the new feature
df[[
    "MonthlyIncome",
    "TotalWorkingYears",
    "IncomePerYearExperience"
]].head()
```

## 7.2 Removing Irrelevant and Redundant Features

Some columns do not provide useful predictive information and can be removed before further preprocessing.

The following columns are removed:

- **EmployeeCount** — constant value for all employees.
- **StandardHours** — constant value for all employees.
- **Over18** — contains only one category and therefore has no useful variation.

## 7.4 Correlation Analysis

Correlation analysis is used to understand the relationships between numerical features in the feature-engineered dataset.

The correlation coefficient ranges from -1 to +1:

- **+1** indicates a strong positive relationship.
- **-1** indicates a strong negative relationship.
- **0** indicates little or no linear relationship.

The correlation matrix and heatmap are used to identify relationships between features and support feature selection decisions.

```python
# Calculate correlation matrix

correlation_matrix = df_fe.corr(
    numeric_only=True
)

correlation_matrix.head()
```

### Correlation Heatmap

The heatmap provides a visual representation of the correlation between numerical features.

Strong positive or negative correlations can indicate that two variables contain related information. These relationships should be considered when selecting features for subsequent machine learning stages.

```python
# Plot correlation heatmap

plt.figure(figsize=(18, 14))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0,
    linewidths=0.3
)

plt.title(
    "Correlation Heatmap — Feature Engineered HR Dataset"
)

plt.show()
```

## 7.5 Highly Correlated Features

To further investigate feature relationships, feature pairs with an absolute correlation of 0.70 or higher are identified.

A high correlation does not automatically mean that a feature must be removed. The relationship and practical meaning of each feature should also be considered.

```python
# Identify highly correlated feature pairs

corr_pairs = (
    correlation_matrix
    .where(
        np.triu(
            np.ones(correlation_matrix.shape),
            k=1
        ).astype(bool)
    )
    .stack()
    .sort_values(ascending=False)
)

high_corr = corr_pairs[
    abs(corr_pairs) >= 0.70
]

print(
    "Highly correlated feature pairs "
    "(|correlation| >= 0.70):"
)

print(high_corr)
```

Features related to employee experience and tenure may show stronger relationships because they measure related aspects of an employee's career history.

The newly created `TenureRatio` and `IncomePerYearExperience` features provide additional information derived from existing employee attributes.

Highly correlated feature pairs were identified using a threshold of |correlation| ≥ 0.70. Correlation alone was not used as the basis for removing features; feature meaning and usefulness were also considered.

## K-Nearest Neighbors (KNN), Stratified 5-Fold Cross-Validation, GridSearchCV, and Random Forest

This section extends the preprocessing notebook for the HR Employee Attrition classification task.

**Required contribution**
- Base model: K-Nearest Neighbors (KNN)
- Validation: Stratified 5-fold cross-validation
- Hyperparameter tuning: `n_neighbors`, `weights`, and `p` using `GridSearchCV`
- Comparison/ensemble model: Random Forest

The model-development section uses the feature-engineered dataset (`df_fe`). KNN scaling is kept inside a scikit-learn pipeline so the scaler is fitted separately within each cross-validation training fold, reducing data leakage. Random Forest is trained on the same features without scaling.


## 1. Prepare features and target

`Attrition` is the binary target. All other columns in `df_fe` are used as predictors. The split is stratified so that the class proportions are approximately preserved in both training and test sets.

```python
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_validate, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    classification_report, confusion_matrix, ConfusionMatrixDisplay,
    roc_auc_score, RocCurveDisplay
)
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Use the feature-engineered dataset created earlier in this notebook.
X = df_fe.drop(columns=["Attrition"])
y = df_fe["Attrition"].astype(int)

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("\nTraining target distribution:")
print(y_train.value_counts(normalize=True).sort_index().rename(index={0: "No", 1: "Yes"}))
print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True).sort_index().rename(index={0: "No", 1: "Yes"}))
```

## 2. Baseline KNN model

KNN predicts a class from the labels of nearby training observations. Since KNN is distance-based, numeric feature scaling is important. `StandardScaler` is therefore included in a pipeline with KNN.

```python
knn_baseline = Pipeline(steps=[
    ("scaler", StandardScaler()),
    ("model", KNeighborsClassifier())
])

knn_baseline.fit(X_train, y_train)
y_pred_knn = knn_baseline.predict(X_test)

print("Baseline KNN Classification Report")
print(classification_report(
    y_test, y_pred_knn,
    target_names=["No Attrition", "Attrition"],
    zero_division=0
))

baseline_accuracy = accuracy_score(y_test, y_pred_knn)
baseline_precision = precision_score(y_test, y_pred_knn, zero_division=0)
baseline_recall = recall_score(y_test, y_pred_knn, zero_division=0)
baseline_f1 = f1_score(y_test, y_pred_knn, zero_division=0)

print(f"Accuracy : {baseline_accuracy:.4f}")
print(f"Precision: {baseline_precision:.4f}")
print(f"Recall   : {baseline_recall:.4f}")
print(f"F1-score : {baseline_f1:.4f}")
```

```python
ConfusionMatrixDisplay.from_predictions(
    y_test, y_pred_knn,
    display_labels=["No Attrition", "Attrition"],
    cmap="Blues"
)
plt.title("Baseline KNN — Confusion Matrix")
plt.tight_layout()
plt.show()
```

## 3. Stratified 5-fold cross-validation for baseline KNN

Stratified K-Fold keeps the proportion of each target class approximately consistent across folds. Cross-validation is performed on the training set only; the test set remains untouched until final evaluation.

```python
stratified_cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_results = cross_validate(
    knn_baseline,
    X_train,
    y_train,
    cv=stratified_cv,
    scoring={
        "accuracy": "accuracy",
        "precision": "precision",
        "recall": "recall",
        "f1": "f1",
        "roc_auc": "roc_auc"
    },
    n_jobs=-1,
    return_train_score=False
)

cv_summary = pd.DataFrame({
    "Metric": ["Accuracy", "Precision", "Recall", "F1-score", "ROC-AUC"],
    "Mean CV Score": [
        cv_results["test_accuracy"].mean(),
        cv_results["test_precision"].mean(),
        cv_results["test_recall"].mean(),
        cv_results["test_f1"].mean(),
        cv_results["test_roc_auc"].mean()
    ],
    "Std. Deviation": [
        cv_results["test_accuracy"].std(),
        cv_results["test_precision"].std(),
        cv_results["test_recall"].std(),
        cv_results["test_f1"].std(),
        cv_results["test_roc_auc"].std()
    ]
})
display(cv_summary.round(4))
```

## 4. KNN hyperparameter tuning using GridSearchCV

The search tests:
- `n_neighbors`: number of nearest neighbors used to vote
- `weights`: `uniform` gives each neighbor equal weight; `distance` gives closer neighbors greater influence
- `p`: distance metric parameter for Minkowski distance (`p=1` Manhattan, `p=2` Euclidean)

The scaler remains inside the pipeline so it is refitted within each CV fold.

```python
knn_pipeline = Pipeline(steps=[
    ("scaler", StandardScaler()),
    ("model", KNeighborsClassifier())
])

knn_param_grid = {
    "model__n_neighbors": list(range(3, 32, 2)),
    "model__weights": ["uniform", "distance"],
    "model__p": [1, 2]
}

knn_grid = GridSearchCV(
    estimator=knn_pipeline,
    param_grid=knn_param_grid,
    scoring="f1",
    cv=stratified_cv,
    n_jobs=-1,
    refit=True,
    verbose=1
)

knn_grid.fit(X_train, y_train)

print("Best parameters:", knn_grid.best_params_)
print(f"Best mean CV F1-score: {knn_grid.best_score_:.4f}")

knn_tuned = knn_grid.best_estimator_
y_pred_knn_tuned = knn_tuned.predict(X_test)
```

```python
print("Tuned KNN Classification Report")
print(classification_report(
    y_test, y_pred_knn_tuned,
    target_names=["No Attrition", "Attrition"],
    zero_division=0
))

ConfusionMatrixDisplay.from_predictions(
    y_test, y_pred_knn_tuned,
    display_labels=["No Attrition", "Attrition"],
    cmap="Blues"
)
plt.title("Tuned KNN — Confusion Matrix")
plt.tight_layout()
plt.show()
```

## 5. Comparison model: Random Forest

Random Forest is an ensemble learning method that trains multiple decision trees and combines their predictions. Unlike KNN, it does not require feature scaling. We use the same training/test split so its results can be compared fairly with baseline and tuned KNN.


```python
random_forest = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)

random_forest.fit(X_train, y_train)
y_pred_rf = random_forest.predict(X_test)

print("Random Forest Classification Report")
print(classification_report(
    y_test, y_pred_rf,
    target_names=["No Attrition", "Attrition"],
    zero_division=0
))

ConfusionMatrixDisplay.from_predictions(
    y_test, y_pred_rf,
    display_labels=["No Attrition", "Attrition"],
    cmap="Blues"
)
plt.title("Random Forest — Confusion Matrix")
plt.tight_layout()
plt.show()
```

## 6. Compare model performance

The following table compares baseline KNN, tuned KNN, and Random Forest on the same held-out test set. Since attrition classes are often imbalanced, do not interpret accuracy alone; precision, recall, F1-score, and ROC-AUC provide additional context.


```python
models = {
    "Baseline KNN": (knn_baseline, y_pred_knn),
    "Tuned KNN": (knn_tuned, y_pred_knn_tuned),
    "Random Forest": (random_forest, y_pred_rf)
}

comparison_rows = []
for model_name, (model, predictions) in models.items():
    row = {
        "Model": model_name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision": precision_score(y_test, predictions, zero_division=0),
        "Recall": recall_score(y_test, predictions, zero_division=0),
        "F1-score": f1_score(y_test, predictions, zero_division=0)
    }
    if hasattr(model, "predict_proba"):
        probabilities = model.predict_proba(X_test)[:, 1]
        row["ROC-AUC"] = roc_auc_score(y_test, probabilities)
    else:
        row["ROC-AUC"] = np.nan
    comparison_rows.append(row)

model_comparison = pd.DataFrame(comparison_rows).set_index("Model")
display(model_comparison.round(4))
```

```python
ax = model_comparison[["Accuracy", "Precision", "Recall", "F1-score", "ROC-AUC"]].plot(
    kind="bar",
    figsize=(11, 6),
    ylim=(0, 1)
)
ax.set_title("Model Performance Comparison on the Test Set")
ax.set_ylabel("Score")
ax.set_xlabel("Model")
ax.tick_params(axis="x", rotation=0)
ax.legend(loc="lower right")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
for model_name, (model, _) in models.items():
    if hasattr(model, "predict_proba"):
        RocCurveDisplay.from_estimator(
            model, X_test, y_test, name=model_name, ax=plt.gca()
        )
plt.plot([0, 1], [0, 1], linestyle="--", label="Random baseline")
plt.title("ROC Curves — Baseline KNN, Tuned KNN, and Random Forest")
plt.tight_layout()
plt.show()
```
